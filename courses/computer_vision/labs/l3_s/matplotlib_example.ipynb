```python
import matplotlib.pyplot as plt
Taipei_temp = [16.1,16.5, 18.5, 17.9, 18.2, 20.7, 21.5, 19.7, 20.2, 20.5, 21.4, 19.4]
Taichung_temp = [12.8, 13.6, 14.8, 17.0, 18.8, 20.0, 21.5, 22.7, 23.1, 22.1, 19.9, 17.6]
Kaohsiung_temp = [25.1,24.8, 25.5, 24.9, 25.2, 27.7, 29.5, 30.7, 30.2, 31.5, 30.4, 29.7]
year = range(2010, 2022)
plt.plot(year, Taipei_temp, color = 'blue', marker = 'o', linestyle= '--',label = 'Taipei')
plt.plot(year, Taichung_temp, color = 'yellow', marker = 'o', linestyle= '-',label = 'Taichung')
plt.plot(year, Kaohsiung_temp, color = 'red', marker = '.', linestyle= '-',label = 'Kaohsiung')
plt.xlim(2009, 2022) 
plt.ylim(10, 35)
plt.legend(loc = 'upper left')
plt.xlabel('Year',color = 'purple')
plt.ylabel('Temperature',color = 'purple')
plt.title('10-year Average Temperature',color = 'purple')
plt.xticks(year,[i for i in range(99,111)])
plt.show() 
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
#建立Figure和Axes物件
fig, ax = plt.subplots()
#primitive artist
scatter_artist = ax.scatter(x, y1, label='Sin')
line_artist = ax.plot(x, y2, label = 'Cos')

#圖例container
legend_artist = ax.legend()

ax.set_title('Sin and Cos Curves')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')

plt.show()
```

```python
fig, ax = plt.subplots(2, 2, sharex = 'col', sharey = 'row')
ax[0, 0].set(title = 'Upper Left', facecolor = 'lightgreen')
ax[0, 1].set(title = 'Upper Right', facecolor = 'lightblue')
ax[1, 0].set(title = 'Lower Left', facecolor = 'coral')
ax[1, 1].set(title = 'Lower Right', facecolor = 'pink')

for axes in ax.flat:
    axes.set(xticks=[], yticks=[])
plt. show()
```

```python
import matplotlib.pyplot as plt
grid = plt.GridSpec(2, 3, wspace = 0.5, hspace = 0.5)
plt.subplot(grid[0, :2], facecolor = 'lightgreen')
plt.subplot(grid[0, 2], facecolor = 'lightblue')
plt.subplot(grid[1, 0], facecolor = 'coral')
plt.subplot(grid[1, 1:], facecolor = 'pink')
plt.show()
```
